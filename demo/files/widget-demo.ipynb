## Codelens cell magic

````python
import codelens_widget
````

````python
%%codelens

def insertion_sort(a):
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return a

data = [5, 2, 9, 1]
insertion_sort(data)
````
