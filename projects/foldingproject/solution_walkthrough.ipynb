## The finished solution

All functions written — every check should pass:

```python
%%test foldingproject
#############################################################
# Reference solution for the folding (primer analysis) project.
# Derived from the project write-up and the stub docstrings.
#############################################################


def count_bases(seq):
    counts = {'A': 0, 'T': 0, 'G': 0, 'C': 0}
    for base in seq:
        counts[base] += 1
    return counts


def melting_temp(seq):
    counts = count_bases(seq)
    a, t, g, c = counts['A'], counts['T'], counts['G'], counts['C']
    if len(seq) < 14:
        return (a + t) * 2 + (g + c) * 4
    else:
        return 64.9 + 41 * (g + c - 16.4) / (a + t + g + c)


def reverse_complement(seq):
    complement = {'A': 'T', 'T': 'A', 'G': 'C', 'C': 'G'}
    result = []
    for base in reversed(seq):
        result.append(complement[base])
    return ''.join(result)


def has_hairpin(seq, k):
    looplen = 4
    for i in range(len(seq) - k + 1):
        subs = seq[i:i + k]
        right = seq[i + k:]
        revcl = reverse_complement(subs)
        if revcl in right[looplen:]:
            return True
    return False
```
