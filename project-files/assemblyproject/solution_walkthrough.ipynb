```python
%%test assemblyproject
#############################################################
# Reference solution for the assembly project.
#############################################################


def read_data(file_name):
    reads = {}
    input_file = open(file_name)
    for line in input_file:
        line = line.strip()
        if not line:
            continue
        name, sequence = line.split()
        reads[name] = sequence
    input_file.close()
    return reads


def mean_length(reads):
    count = 0
    total = 0
    for name in reads:
        total += len(reads[name])
        count += 1
    return total / count


def get_overlap(left, right):
    for size in range(min(len(left), len(right)), 0, -1):
        if left[-size:] == right[:size]:
            return left[-size:]
    return ""


def get_all_overlaps(reads):
    overlaps = {}
    for left_name in reads:
        overlaps[left_name] = {}
        for right_name in reads:
            if left_name == right_name:
                continue
            overlap = get_overlap(reads[left_name], reads[right_name])
            overlaps[left_name][right_name] = len(overlap)
    return overlaps


def pretty_print(d):
    print('      ', end='')
    for j in sorted(d):
        print("{: >6}".format(j), end='')
    print()

    for i in sorted(d):
        print("{: >6}".format(i), end='')
        for j in sorted(d):
            if i == j:
                s = '     -'
            else:
                s = "{: >6}".format(d[str(i)][str(j)])
            print(s, end='')
        print()


def get_left_overlaps(overlaps, name):
    left_overlaps = []
    for other_name in overlaps:
        if other_name == name:
            continue
        left_overlaps.append(overlaps[other_name][name])
    return sorted(left_overlaps)


def find_first_read(overlaps):
    for name in overlaps:
        left_overlaps = get_left_overlaps(overlaps, name)
        if max(left_overlaps) <= 2:
            return name


def find_key_for_largest_value(d):
    best_key = None
    best_value = None
    for key in d:
        if best_value is None or d[key] > best_value:
            best_value = d[key]
            best_key = key
    return best_key


def find_order_of_reads(first, overlaps):
    order = [first]
    current = first
    for _ in range(len(overlaps)):
        next_read = find_key_for_largest_value(overlaps[current])
        if overlaps[current][next_read] > 2:
            order.append(next_read)
            current = next_read
        else:
            break
    return order


def reconstruct_sequence(order, reads, overlaps):
    sequence = reads[order[0]]
    for i in range(1, len(order)):
        overlap = overlaps[order[i - 1]][order[i]]
        sequence += reads[order[i]][overlap:]
    return sequence


def assemble_genome(file_name):
    reads = read_data(file_name)
    overlaps = get_all_overlaps(reads)
    first = find_first_read(overlaps)
    order = find_order_of_reads(first, overlaps)
    return reconstruct_sequence(order, reads, overlaps)
```
