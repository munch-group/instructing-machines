You may wonder why the arguments to the `sns.displot` function are written as name=value rather than just the values. 

In Python, you can pass values to a function in this way too. Such keyword arguments (often called named arguments) explicitly stating the parameter name and its value using the parameter_name = value syntax. This is useful for three reasons:

- Order independence: Keyword arguments do not need to match the exact order of parameters defined in the function. This is useful if there are many in an order that hard to remember.
- Readability: Code becomes self-documenting because the intent behind each value is explicitly labeled.
- Selective defaults: If a function has multiple parameters with default values, you can override just one specific parameter without passing values for the others too.
