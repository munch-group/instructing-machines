# AI: Plan before you prompt {#sec-plan-before-you-prompt}

By now you know how you can use a set of tests as a specification of how a function should work. The function meets its specification only if it passes all the tests. The nice thing about functions is that, once they are made, you no longer have to care how they work. You just need to which types of values they take as arguments and what type of value they return. This "interface" with the larger program is called the function's signature. Once you have decided on a functions signature and you have written a slew of tests that define what it should return across an exhaustive range of different inputs, it *almost* does not matter what is in the function body. It may be implemented in any what that respects the signature and passes all the tests. Having defined what constitutes a correct function body ahead of time, makes it safe to delegate the function body an AI. 

::: {.column-margin}

A function's arguments and return value is referred to as its *signature* and defines its interface to the larger program.

:::

This approach is also how you stay in control when you have the AI help you write a complex program. If you decided to  cram the whole thing into a a single function, you would need to write tests that cover all the variations and corner cases of inputs and return values. It would not only be difficult, it would also be hard to assure yourself that you in fact did cover all the cases. And not being completely sure, kind of defeats the purpose of doing the tests in the first place. Instead, you want to break the solution to the problem into smaller functions, each with a defined signature and tests. The smaller each function is, the easier it is to reliably write a set of exhaustive tests. What that means in practice, is that you need to break down the problem yourself *before* asking the AI to fill in all the function bodies. Fortunately, this is often not that difficult if you understand the problem you are trying to solve. It will also make you realize if you, in fact, do *not* understand the problem well enough to reliably test result the AI would hand bck to you. This breaking the problem down into functions is called *decomposition*. You have seen examples of this in the programming projects you have completed so far. 

The final product of your decomposition is really just a list of function stubs, with tests that should pass once their bodies have been filled in. Python *docstrings* will help with your planning for for remembering what each function does. Docstrings are just strings you put on the first line of the function. If formatted correctly, they can be used to generate online documentation, but for our purposes they just contain your own definition of what each function does.

I have only included tests for the `gc_content` function here but I am sure you get the idea. You can also add them to a single notebook cell like I did here and use the `%%test` magic to run the tests (once you have imported `im_pytest`). The tests will keep failing until all function bodies are filled correctly.
