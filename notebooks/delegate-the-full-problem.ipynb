# AI: Delegate the full problem {#sec-delegate-the-full-problem}

This last level on the AI ladder, is not a badge you should pin on your vest just yet. This course has been short. Before you begin handing over the entire task to the AI, you need a lot more experience using the safe approach introduced in @sec-plan-before-you-prompt. That said, I know you will head off to fight Darth Vader before your jedi training is complete. So before you board your X-wing, here are some pointers that I hope you follow will.

The only way you can stay in control of your work with an AI, is to never ask it for something you do not know what is. Use the same steps that you have learned along the way, but delegate each one to the AI in separate steps. 

Make sure you understand the problem you are going to solve. Begin by outlining to the AI, what the problem you want solved is. Say that you are not asking for code, but want to make sure you agree what the problem to solve is. Have it repeat back to you how it understood the problem you explained. Ask it for a Q and A, probing any unclear points and corner cases:

::: {.prompt}
> I need code that finds open reading frames (ORFs) in DNA sequence. Do not produce code until I way so. We need to first agree the problem at hand. Please quiz me on what the final product should look like.
:::

When you are sure you agree on wha the problem is, make sure you agree on how to solve it. Say that you want it to it decompose the problem into functions explained by their docstrings and that these functions should be so small that each one can be tested reliably:

::: {.prompt}
> Please suggest how the required code can be decomposed into functions small enough to be tested reliably. Show me as function stubs with short docstrings.
:::

Read and approve or ask it to revise the decomposition it suggests. When the decomposition is in place, ask it to write *pytest* tests for each function just like we did in @sec-plan-before-you-prompt:

::: {.prompt}
> Please write test functions in the same file to be run with pytest. 
:::

Read the tests to make sure the tests cover the relialistic use-cases you have in mind. Then ask it to fill in the bodies and run the tests you on the finished functions:

::: {.prompt}
> Now please fill in the function bodies and run the test suite on the finished functions.
:::

Review the code to make sure you understand how it works. If you do not, ask the AI to explain it to you and/or write out the code more explicitly:

::: {.prompt}
> I don't understand the code in the last function. Please add inline comments to explain each step of the code.
:::
