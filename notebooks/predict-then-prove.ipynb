# AI: The AI predicts, you prove {#sec-predict-then-prove}

*A claim about what code does is a prediction to be verified.*

The routine for this whole session is the same three-way move, repeated on harder and harder expressions until it is automatic. You predict what an expression evaluates to, by hand, doing the substitution and reduction steps yourself. You ask the assistant what it evaluates to, and read its answer, including whatever reasoning it gives. Then you run it through the `%%steps` widget, which is the only one of the three that actually reduces the expression step by step and cannot be talked around. Three answers, one witness. When your prediction and the assistant's answer agree, that is pleasant and proves nothing. When they disagree, or when either of you turns out to be wrong, that is the session working exactly as intended.

::: {.callout-tip}
## The three-way move, every time
1. **Predict.** Write down your answer and the substitution/reduction steps that got you there, before you type anything into the assistant.
2. **Ask.** Give the assistant the bare expression and ask what it evaluates to and why. Read its reasoning, not just its final answer.
3. **Prove.** Run the expression through `%%steps`. Its trace is the verdict. Compare it against both of your answers and note any disagreement.
:::

Every round below carries the `AI: Explainer` badge, which is the rung of the ladder you were given in week one and the only one you have so far. You are putting something in front of the assistant that already exists and asking it to explain that thing; you are not asking it to write anything, and you are not asking it to choose between two versions. What makes this session demanding is not the permission, which is the smallest one there is, but how hard you are going to work it. level one, twenty-two times, until checking a claim against a machine is what your hands do without being told.
