## Commands and questions {#sec-turtle-week-3-commands}

The turtle methods can be divided into to kinds: The first are the ones that make the turtle do something, thereby changing the state of the turtle object in terms of location, direction, etc. (like `forward`, `backward`, `left`, `right`, `penup`, `pendown` and `goto`). The second are are methods for extracting information about the state of the turtle object: where it is, which way it points etc. (like `position`, `xcor`, `ycor`, `heading`, `isdown` and `isvisible`).

## Where is my turtle {#sec-turtle-week-3-questions}

The turtle starts out in the middle of the canvas facing right. Every position on the canvas as an x,y coordinate with x values growing to the right and y values growing upwards. A heading of zero means the turtle is facing right and the heading grows as the turtle turns counter-clockwise. So a heading of 90 is straight up.

## Two names, same turtle {#sec-turtle-week-3-aliasing}

This week you saw that giving a list a second name does not give you a second list. Turtles behave the same way, and because a turtle is visible the demonstration is unusually blunt.

#### Exercise

[SOLO](../intro/course-introduction.qmd#sec-badge-solo){.small}

Decide what the drawing will look like and what the last code line prints. Then run it.


```python
from turtle_widget import Turtle

t = Turtle()
other = t
other.forward(100)
other.left(90)
other.forward(100)
print(t.position())
```

Everything you did using `other` happened to `t`, because `other` and `t` are the same thing. It works the same as it does with a variable pointing to a list: `a = [1, 2, 3]` followed by `b = a` makes `b` point to the same list as `a`. `t` points to an object of type `Turtle` so `other = t` makes other point to whatever `t` points to.
