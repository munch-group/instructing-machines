As you learned in @sec-objects, python values are objects, which package the value or data with useful functionality in the form of *methods*. Methods are very similar to functions but know what data they are packaged with. So calling the `upper` method on a sting (`"banana".upper()`) amounts to saying: "*Hey string, uppercase yourself!*". If you had a function that could uppercase a string, calling it would be like saying "*Hey function, uppercase this string*". 

````python
%%sandbox
"banana".upper()
````

Now you may ask: where does `"banana"` keep that `upper` method? The string `"banana"` is stored in your computers memory along with all the other strings you have defined. You may think that each one somehow carries around the code defining the definitions of `upper`, and `lower`, and `strip`, and `replace`, and the forty-odd other string methods? That would be an awful lot of luggage for each object. If each object carried its own method definitions, it would take up a lot of memory on top of the Python value it carries and would fill up memory with duplicate definitions as you define more strings. To avoid this, all objects of the same *type*, say strings, share their method definitions and everything else that goes into defining an object of that type. That way, storing an extra string like `"banana"` only takes six bytes of extra memory.

In @fig-objects_and_type you can see how individual objects relate to the type that defines them. The two string objects carry only their data (in this case `"Preben"` and `"Mogens"`) and the knowledge that they are of type `str`. They are different objects and are stored in different places in memory, but defer to a single shared definition of their methods.

![Objects and type. Each object carries its own data, but refers to the type for a definition of its methods.](./images/objects_and_type.png){#fig-objects_and_type width=50%}

Each individual object holds its own *data*, and the type holds the *functionality*. When you write `"banana".upper()`, Python checks which type the object is, finds it is a `str`, gets the `upper` method from the `str` type, and runs it on the particular (banana) data carried by the object.

````python
%%sandbox
type("banana")
````

## Classes define types

The type of an object is defined by its *class*. The `str` type is defined by the `str` class. So far you have seen the builtin types `str`, `int`, `float`, `bool`, `list`, `tuple`, and `dict`. There are many more, and the beauty of it is that you can make them yourself too. To do that, you write a class that defines your type of object. Populating your class with definitions of methods makes these available to every object of that type.

What might a useful type of object to make? It would carry some particular kind of data that it makes sense to pair with some custom functionality. Could it be a kind of biological data? Maybe an aggregate of data values that together represent something meaningful? Ponder on that for a bit before you go on?

To create a new point, you simply call the `Point` class using the round parentheses as you would a function:

````python
p = Point()
print(p)
````

At this point our class does not define any data or functionality. So the object it creates is no different from the generic `object` that all new objects start out as. We need to *initialize* the generic `object` to make it our own. We do that by adding a a special method called `__init__` that runs when an object is created:

````python
class Point(object):
    def __init__(self):
        print("Yay! I exist. My type is", type(self))

p = Point()
````

As you can see, a method definition is just a function definition inside a class. Everything you know about functions definitions apply to methods as well. The `self` parameter points to the newborn object to be initialized.

In the version above, the `__init__` method just prints a happy message. We would like `Point()` to accept two arguments like `Point(2, 5)`  To do that we just add `x` and `y` parameters to the method:

::: {.column-margin}

You may wonder where the value of `self` comes from since we did not pass any arguments when we call `Point()`. The reason is that when you call `Point()`, what really happens is `Point.__init__(object)`.

:::

````python
class Point(object):
    def __init__(self, x, y):
        print("Yay! I have coordinates: {},{}".format(x, y))

p = Point(2, 5)
````

Now that we can pass coordinates to the new object, we just need it to also store them as its data. We store each one as object *attributes*  using the `.` operator:

````python
class Point(object):
    def __init__(self, x, y):
        self.x = x
        self.y = y
````

Now the created point object has `x` and `y` *attributes*, which are basically object variables like any other that we can access using the `.` operator. The dot is the same one you have used to access methods. It basically means: *go into this object and get me the thing called...*. That thing may be a method and it may be an attribute storing some data value.

````python
p = Point(2, 5)

print(p.x)
print(p.y)
````

## How should it look as a string?

Printing a point gave us that unfriendly `<__main__.Point object at ...>`. Let us do something about that. When you print an object, Python needs to turn the object into some kind of meaningful string to print. If the type of object has a `__str__` method, it will use whatever that method returns. If not, it will fall back to `<__main__.Point object at ...>`. 

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return "({}, {})".format(self.x, self.y)


p = Point(4, 3)
print(p)
````

See how nicely it looks now. We could also have made it `"Point({}, {})".format(self.x, self.y)`. What would that have printed?

::: {.column-margin}

`self` is just the object *itself*. Again, if you wonder where that value comes from the `self`, it may make more sense if you know that what really happens when you `print(p)` is `print(Point.__str__(p))`.

:::

If you put `p` on the last line of a cell and expect the value to display as `(4, 3)`, you will be disappointed.

````python
p
````

This it is because displaying the value produced by the last line of code cell does not call `__str__`, it calls the one named `__repr__`. So if you want to fix this, all you have to do is make a copy of your `__str__` method and call it that `__repr__`:

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return "({}, {})".format(self.x, self.y)

    def __repr__(self):
        return "({}, {})".format(self.x, self.y)

p = Point(4, 3)
p    
````

## A method of your own

Now lets us add a method that actually computes something. How far is a point from origin? That is the Pythagorean theorem you used in @sec-pythagoras: the distance is $\sqrt{x^2 + y^2}$, and taking a square root is exponentiation to $0.5$.

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return "({}, {})".format(self.x, self.y)

    def __repr__(self):
        return "({}, {})".format(self.x, self.y)
    
    def dist_from_origin(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5
````

Notice how we don't need to pass any arguments to `dist_from_origin` because all the information is already carried by the object. Read it and run it. You should get `5.0`, which is the nicest right-angled triangle there is.

````python
p = Point(4, 3)
dist = p.dist_from_origin()
print(dist)
````

Every point you ever make from now on can do this. @fig-point_and_type_methods is the same picture as before, now with your own type in the blue box.

![Two point objects and their type. The data lives in the objects; the methods live in the type.](./images/point_and_type_methods.png){#fig-point_and_type_methods width=75%}

## Peeking behind the curtain, again

You have already added three of the special methods with double underscores around their names, methods that Python calls when it needs to. There are many more, each one defining behavior of the object in some context. 

Remember `__add__` from @sec-add_method? The `+` operator does not know how to add anything. It asks the object on its left to add the object on its right, by calling its `__add__` method. So if you give your class an `__add__` method, `+` will work on your objects too:

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return "({}, {})".format(self.x, self.y)

    def __repr__(self):
        return "({}, {})".format(self.x, self.y)
    
    def dist_from_origin(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5

    def distance(self, other):
        ...
        ...
        return ...

    def __add__(self, other):
        new_x = self.x + other.x
        new_y = self.y + other.y
        return Point(new_x, new_y)
````

Look carefully at what `__add__` returns. It is not a number but a brand new point `Point(new_x, new_y)`. A class can make objects of its own type inside its own methods. 

````python
p1 = Point(4, 2)
p2 = Point(3, 7)
print(p1 + p2)
````

## A Gene class

A gene might hold a DNA sequence, and asking how long a gene is should be an easy thing to do. When you use the builtin `len` function on an object, Python calls the objects `__len__` function if it exists. Have a look at this one:

````python
%%sandbox
class Gene(object):
    def __init__(self, name, sequence):
        self.name = name
        self.sequence = sequence

    def __str__(self):
        return "{} ({} bases)".format(self.name, len(self.sequence))

    def __len__(self):
        return len(self.sequence)


gene = Gene('BRCA1', 'ATGGATTTATCTGCTCTTCGCGTT')
print(gene)
print(len(gene))
````

Notice that `__len__` does not count anything itself. It hands the question on to the string it is holding, because a string already knows how long it is. Most of the methods you write will be like this: short, and mostly delegating to objects that already know how to do the work.
