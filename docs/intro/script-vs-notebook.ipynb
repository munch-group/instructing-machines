# Scripts and notebooks {#sec-script-vs-notebook}

You have written a program, saved it in a file, and run it from the terminal. That is one way of running Python, and it is the way most Python in the world gets run. This chapter shows you the other way, the one you are looking at right now, because this page is itself a running Python program. The two ways are not the same, and the places where they differ are exactly the places where beginners get confused. So let us take them apart.

## The script you already know {#sec-the-script}

Your `hello.py` is a script. It is a plain text file sitting on the disk with some Python code written in it. To run it, you go to the terminal and type:

```{.txt filename="Terminal"}
python hello.py
```

The program called `python` is the interpreter. It opens your file, reads the first line, does what it says, reads the next line, does what that one says, and carries on until there are no more lines. Then it stops and exits.

Two things follow from that, and both of them matter more than they look.

The first is Oath 2 again: the order the lines appear in the file is the order in which they run. Nothing jumps around. If you want to know what a script does, you read it downward.

The second is that every run starts from nothing. When the interpreter exits, everything it was holding is thrown away, because it was all in memory, and memory is emptied the moment a program ends. Run the script again tomorrow and it starts from a blank slate once more. A script therefore carries everything it needs inside itself. If it works today, it works tomorrow, on your machine or on someone else's, and that reliability is bought entirely by starting from nothing every single time.

## The notebook you are reading {#sec-the-notebook}

Now look at where you are. This page is a notebook, a file whose name ends in `.ipynb`. You can download it using the link in the margin and open it in VScode. Instead of one long stream of code, a notebook is built from cells. Some cells are text, like this one. Other cells hold code and can be run one at a time.

Enough description. Run the cell below. Click it and press Shift and Enter together.

````python
dna = "ATGGTGCATCTGACTCCTGAGGAGAAG"
len(dna)
````

Two things just happened. A small number appeared in brackets to the left of the cell, and the length of the string appeared underneath it even though you never wrote `print`. Hold on to both of those; we will come back to them.

Now run the next cell.

````python
dna.count("A")
````

That cell never mentions what `dna` is, and yet it worked. This is the whole difference in one line. Behind this page there is a Python process that has been running the entire time, and it is still running now. It is called the kernel. When you run a cell, the code in that cell is handed to that same living process, which does what it says and then waits for whatever you hand it next. The first cell put a string into that process and gave it the name `dna`. The second cell asked the same process about a name it already knew.

So a notebook is not a file that the interpreter reads from top to bottom. It is a pile of small pieces of code that you hand, one piece at a time, to a session that remembers everything you have handed it so far.

The number in brackets is the execution count, and it tells you the order in which you handed things over. The first cell you run in a session gets a `1`, the next gets a `2`, and so on. Run the same cell again and its number goes up. This is the notebook quietly telling you something important, which is that the kernel remembers the order you ran cells in, not the order the cells sit on the page. You can run the bottom cell first, then the top one, then the bottom one again, and the kernel will happily oblige. You can even delete a cell after running it, and the names it defined will still be sitting in the kernel, alive and usable, with nothing on the page left to explain where they came from.

That is the notebook's great convenience and its great trap. It is a convenience because you can try one small thing, look at the answer, and try the next small thing, without rerunning your whole program each time. It is a trap because what you see on the page can stop matching what the kernel actually holds.

The other loose end was the length appearing without `print`. When the last line of a cell is a bare expression, something that reduces to a value, the notebook displays that value for you. Only the last line, and only if it is an expression. An assignment like `dna = "ATG"` produces no value, so nothing is shown. This is a courtesy of the notebook, not a rule of Python. A script will print nothing at all unless you ask it to with `print`, which is one of the small ways that code moved out of a notebook stops behaving the way you remember.

## Restart and run all {#sec-restart}

The exercise you just did is the single most useful habit you can form about notebooks, so let us give it a name. Restarting the kernel throws away the running process and starts a fresh one that knows nothing. Running all the cells afterward feeds them back in, from the top, in page order. In other words, restarting and running all makes your notebook behave, for one run, like a script.

Do this before you believe a notebook works, and certainly before you hand one to anybody else. A notebook that only works because of something you ran twenty minutes ago in a cell you have since deleted is a notebook that will fail for the next person who opens it, and that next person is often you, next week.

## Python and IPython {#sec-ipython}

There is a second thing going on that you have not been told about yet, and now is the time.

If you type `python` in the terminal with no file name after it, you do not get an error. You get a prompt that looks like this:

```{.txt filename="Terminal"}
Python 3.13.0
>>>
```

That is the interpreter waiting for you to type one line of Python at a time. You type a line, it runs it, it shows you the answer, and it waits again. It is the same interpreter that runs your scripts, just taking its instructions from your keyboard instead of from a file. Type `exit()` to leave.

There is also a second such prompt, from a program called IPython, which you start by typing `ipython`. It looks like this:

```{.txt filename="Terminal"}
In [1]:
```

IPython is not a different language. It is the same Python, with a friendlier shell wrapped around it: it completes names when you press Tab, it colors your code as you type, it prints much nicer error messages, and if you put a question mark after a name it shows you the documentation for it. It also adds something Python itself does not have, which we are about to meet.

And here is the point of this section. The kernel behind a Python notebook is IPython. That bracketed number next to your cells is IPython's counter, the same one that says `In [1]` at its terminal prompt. You have been using IPython since the first cell of this chapter without knowing it.

### Magic commands {#sec-magics}

Run this cell.

````python
%whos
````

What you just ran is not Python. It is a magic command, one of IPython's own commands, and `%whos` asks it to list every name the kernel is currently holding. Look at that list and notice that `dna` is in it. That table is the kernel's memory made visible, and it is a good thing to run whenever a notebook starts behaving strangely.

Magic commands come in two shapes. A line magic starts with one percent sign and affects only its own line, so `%whos` above is a line magic. A cell magic starts with two percent signs, must be the very first line of the cell, and takes charge of everything below it in that cell. Every widget in this course is a cell magic, which is why they all look like `%%steps`, `%%puzzle`, `%%codelens` and, in a moment, `%%sandbox`.

Because magics belong to IPython and not to Python, they only work where IPython is running, which means in a notebook or at the `ipython` prompt. Copy `%whos` into `hello.py`, run `python hello.py`, and you will get a `SyntaxError`, because the plain interpreter has never heard of it. Keep that in mind the next time you move code out of a notebook and into a script and it stops working for no visible reason.

## When the kernel's memory gets in the way {#sec-in-the-way}

The kernel's long memory is a gift when you are exploring and a menace when you are learning.

Suppose an exercise asks you to run a piece of code that should fail, so that you can read the error message. If you happen to have defined that name in some other cell half an hour ago, the code will not fail at all, and the lesson evaporates. Or suppose you write a small piece of code, run it, see it work, and conclude that it is correct. It may be correct. It may also be leaning on a name that a cell somewhere above quietly left lying around, in which case it will collapse the moment it is moved anywhere else.

What we want, for a lot of the exercises in this book, is a cell that behaves like a script: one that starts from nothing, contains everything it needs, and leaves nothing behind. That is what the sandbox is for.

## The sandbox {#sec-sandbox}

Run this cell once. It brings in the tool and registers the magic with IPython.

````python
import sandbox_widget
````

Now run this one.

````python
%%sandbox
print("Hello from a fresh interpreter")
````

The output turned up in a shaded box below the cell instead of in the usual place, which is your visual cue that this cell was not run the way the others were.

Here is what makes it different. Predict what the next cell will do before you run it. Remember that `dna` has been sitting in the kernel since the top of the chapter, and that the cell below asks for it by name. This cell is meant to fail.

````python
%%sandbox
print("The sandbox is about to look for dna.")
print(dna)
````

It printed the first line and then raised a `NameError` on the second, and both the printed line and the full traceback landed in the box. The notebook itself carried on untroubled.

Now name what you saw. `%%sandbox` is a cell magic that takes everything below it and runs it in a brand new Python interpreter, a separate process started for that one cell and thrown away the moment it finishes. Nothing that the notebook knows about is visible inside it. The kernel has a `dna`, and the sandbox has never heard of it, because the sandbox is not the kernel.

The barrier runs both ways. Run the next two cells in order.

````python
%%sandbox
codon_count = 64
print(codon_count)
````

````python
print("codon_count" in globals())
````

The sandbox cell made a `codon_count` and printed it quite happily. The ordinary cell that followed asked the kernel whether it has ever heard of that name, and the answer is `False`. Whatever the sandbox does stays in the sandbox, and when its process ends, all of it goes.

So a `%%sandbox` cell is a tiny script that happens to live inside a notebook. It runs from nothing, top to bottom, and it has to carry everything it needs, including its own `import` lines. Anything it defines dies with it, and any mistake it makes is caught and shown to you rather than left to poison the rest of the session.

That last part is why so many exercises in this book use it. A cell that is supposed to break can be relied upon to break, every time, for everybody, no matter what you ran before it.

## The other magics in this book {#sec-other-magics}

Every widget you will meet in this course works the same way as the sandbox: you import it once near the top of the notebook, and after that a cell magic is available for the rest of the session. `%%steps` shows you the substitution and reduction steps inside a single statement. `%%puzzle` gives you shuffled lines to put back in order. `%%codelens` walks through a piece of code one statement at a time and draws what is in memory as it goes. And `%%sandbox` runs a cell as though it were a script.

If a magic reports that it does not exist, the answer is almost always that you have not run its import in this kernel, or that you restarted the kernel since you did.

## Which one to use, and when {#sec-which}

Use a notebook when you are exploring, when you want to see an answer before deciding on the next step, and when the text and the code belong together, which is why the notes in this book are notebooks. Use a script when you want something to run the same way every time, when you want to hand it to somebody else, and when the point is the result rather than the journey.

And use the two habits from this chapter to keep yourself out of trouble. When you want to know whether a notebook really works, restart the kernel and run everything from the top. When you want to know whether a piece of code stands on its own, put it in a `%%sandbox` cell and watch it either run or fall over.

That second habit is going to matter a great deal once the assistant starts writing code for you. Code pasted into a notebook that appears to work may only be working because of something the kernel already had lying around from an earlier cell, and that is a very easy way to accept a piece of code that is broken. Dropping it into a sandbox forces it to prove that it stands up by itself. The assistant predicts that its code works. The sandbox is one of the ways you make it prove it.
