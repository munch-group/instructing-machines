# Finale build {#sec-finale-build}

You arrive at this point holding a plan and a first set of tests for a problem you chose yourself. Nothing about the building that follows is new. It is the loop you have already run many times, applied now to your own decomposition rather than to one somebody handed you, and repeated once for every piece in your plan. What is different is that nobody will tell you which piece to build next, nobody will tell you when a piece is finished, and nobody will notice on your behalf when an answer is quietly wrong. Those three judgments are the whole of the work, and they are yours.

## Take the pieces in dependency order {#sec-dependency-order}

Start with a piece that depends on nothing else. In almost every plan there is at least one, usually a small computation over a value passed straight in, and it is the right place to begin because when it fails there is nowhere else for the fault to hide. Once it is proven you build the next piece on top of a foundation you already trust, and you keep doing that until the piece that faces you, the one that reports or plots or prints, is the last thing standing.

Resist the temptation to build the interesting piece first. The interesting piece is usually the one furthest up the dependency chain, and building it early means testing it against pieces that are not yet proven, which means that when something goes wrong you cannot tell whose fault it is. Build boring first. The whole reason the loop stays comfortable is that at any moment exactly one thing in your program is unproven, and it is the thing you are looking at.

## The cycle for a single piece {#sec-single-piece-cycle}

For each piece the cycle is the same five steps, in the same order.

You write the contract as tests, seeded from the hand-worked example already sitting in your plan's docstring, and stiffened with at least one awkward case. You prompt the assistant for that one function and nothing else, giving it the function's purpose, its inputs and outputs, and the tests it has to satisfy. You read what comes back against your judging questions, and you refuse to run anything you cannot read. You run the tests and let the machine deliver the verdict. And when the tests pass, you write the journal entry for that piece before moving on, because a failure you fixed ten minutes ago is already half forgotten.

The order matters at both ends. Tests before the prompt, so that the assistant is aiming at a target you set rather than one it invents. Journal after the pass, so that the record is written while you still remember what actually went wrong rather than reconstructed at the end of the week from a program that now works.

## Prompting for one piece and only one piece {#sec-scope-the-prompt}

The most common way a build clinic goes sideways is that the assistant is allowed to run ahead. You ask for one function and it returns that function plus two helpers it decided you needed, or plus the caller that ties three pieces together, or plus a rewritten version of a piece you had already tested and proven.

Do not accept that. A function you did not plan is a function you did not decide you needed, and every line of it is code you now have to read, understand, test and stand behind, bought at a price you did not agree to pay. If the assistant hands you more than you asked for, take the piece you wanted, discard the rest, and prompt again for the next piece when you actually reach it. If while doing that you find yourself thinking that one of the extra pieces was in fact a good idea, that is a change to your plan, and it belongs in the plan first, with a name and stated inputs and outputs and a hand-worked example, before you accept a line of its code.

## When a piece will not converge {#sec-build-not-converging}

You write a precise change request, naming the failing input, the wrong output, the right output, and the reason as far as you can see it. The next version should be closer. If two or three rounds of that leave you no nearer, stop and read the situation rather than prompting harder.

Usually one of two things is true. Either your specification is wrong or incomplete, and the assistant is faithfully hitting a target you did not mean, in which case the thing to fix is the test and not the code. Or the piece has become small and familiar enough that you should simply write it yourself, because you now understand it well enough that delegating costs more than doing. Taking the keyboard back is a decision, not a defeat, and it is worth recording in the journal as one.

## Verification follows authorship {#sec-authorship-split}

Keep the two kinds of checking apart as you build, because they are not the same activity and confusing them is how projects end up both over-tested and under-verified.

Code you wrote, or directed the assistant to write for you, gets tests. It is yours, you are responsible for it, and a test is how you discharge that responsibility. Output that came out of a library gets a sanity check against a value you already know. You do not write tests for BioPython's genetic code or for a dataframe's grouping. You take one small case whose answer you can work out yourself, run it through the library, and confirm the library agrees with you.

The failure that this split is designed to prevent is subtle. It is the one where a function of yours produces a wrong answer, but a trusted library was involved somewhere inside it, and the presence of the trusted name quietly persuades you that the whole thing must be sound. The library being correct tells you nothing about whether you called it with the arguments you meant, on the strand you meant, in the frame you meant. Your code around the library is still your code.

## The journal as you go {#sec-journal-during-build}

The prompt journal is not a report you write at the end. It is a running record kept while you build, piece by piece, of what you asked for, where the answer was wrong, how you caught it, and what you did about it. Written as you go it costs a minute per piece. Reconstructed afterwards it is guesswork, and it will be visibly guesswork.

The entries that matter most are the ones where something was wrong and you found it. A build in which the assistant never made a mistake is not a build that went unusually well, it is a build in which the checking was not sharp enough to find the mistakes, and a journal recording no failures says more about the verification than about the assistant.

#### Exercise {#sec-ex-build-first-piece}

[AI: Developer](../intro/course-introduction.qmd#sec-badge-developer){.small}

Build the first piece in your dependency order, all the way through the five-step cycle, and do not move on until its tests pass and its journal entry is written. What you hand the assistant is the piece as your plan states it: the name, the inputs and outputs, and the hand-worked example, and nothing else. What you judge is not how the code reads but whether it passes the tests you wrote before you asked. Then say, in one sentence, what you would have to change if the piece turned out later to be wrong, and notice whether that sentence is short because the piece is well isolated or long because it is not.
