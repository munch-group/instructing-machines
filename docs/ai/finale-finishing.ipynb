# Finale finishing {#sec-finale-finishing}

At this point the pieces of your tool exist, each of them tested, each of them built on pieces that were tested before it. What remains is the part that most projects skimp on, which is the finishing: making the whole thing work end to end, testing the one function you have been quietly avoiding, and then stepping back far enough to see what you have actually demonstrated. That last step is not decoration. The point of the finale was never the tool. It was to find out whether you can produce something correct, understood, and defensible in a domain you know, using a machine you do not fully trust, and the finishing is where you find out.

## The function that faces you {#sec-interface-function}

Every tool has a piece at the top that faces the person using it, the one that prints the report, draws the plot, writes the output file, or answers the question the tool was built to answer. It is almost always the last one built and almost always the least tested, for a reason that sounds convincing and is not: you can see it working. It ran, output appeared, the output looked like a report, and so it must be fine.

Look at what that reasoning actually covers. It confirms that the function produced output without crashing. It says nothing about whether the numbers in that output are the right numbers, whether the labels are attached to the values they describe, whether a filter that was supposed to exclude something actually excluded it, or whether a count is off by one at the edges. Every one of those failures produces output that looks exactly like success, because looking right is a property of the formatting and not of the content.

So test it like anything else. Give it an input small enough that you can write the entire correct output by hand, run it, and compare character by character or value by value. If the function prints, capture what it prints and compare it against the exact string you expect. If it plots, arrange for it to also return the values it is plotting and test those, because a plot you can only check by squinting at it is not a plot you have checked. If it writes a file, write a tiny input, run it, read the file back, and compare it against what you wrote out by hand beforehand.

The awkward inputs deserve the same attention here as anywhere else, and at the interface they are usually about emptiness and about the edges. What does your tool report when the input contains no rows that pass the filter? When the input is empty entirely? When two results tie for the answer you asked it to report? Each of those is a decision, and if you do not make it deliberately your tool will make it for you, silently, at the worst possible moment.

## Running it end to end on real data {#sec-end-to-end}

With every piece tested individually, run the tool on the real input it was built for, and then do the thing that separates a finished tool from one that merely runs. Take one result out of the output and verify it by hand, independently, from the original data.

Pick a row, a gene, a sequence, whatever unit your tool reports on, find it in the source data yourself, compute the answer your tool claims for it using nothing but your own head and perhaps a calculator, and check that the two agree. If they do, you have evidence that the pieces are not only individually correct but correctly connected, which is a different claim and one that unit tests on the pieces cannot make for you. If they disagree, you have found an integration fault, and you have found it now rather than after presenting the result.

Do this at least twice, and choose the second case deliberately from somewhere odd: the shortest sequence, the row with a missing value, the entry at the boundary of whatever your tool filters on. The middle of the data is where everything works.

## What you actually built {#sec-augmentation}

Now step back. Look at the finished tool and ask honestly who did what.

You brought the biology. You knew which question was worth asking, what the data meant, which answer would be suspicious, and what an off-by-one on a coordinate or a read of the wrong strand would do to the result. Nobody supplied that, and the assistant could not have. You also brought the structure, because you decided what the pieces were, what each one promised, and how they fitted together, and you wrote that down before anything was typed.

The assistant brought fluent typing of pieces you had already specified. It filled in shapes you drew. On several of them it was wrong, and your journal records where, and the tests or your own reading caught it. That is the honest accounting, and it is worth stating plainly because it is neither of the two stories people usually tell. It is not that you wrote a program with a bit of help, and it is not that the machine wrote your program. You produced something correct that you could not have typed unaided, by knowing what correct meant and refusing to accept anything you had not checked.

That is augmentation, and it is the thing this course was actually about. A machine that types faster than you is only useful to someone who can tell right from plausible. The reason the term spent so long on what a program is, on substitution and reduction, on reading errors, on writing tests, was never that you would need to write everything yourself. It was that none of the checking works without it. The people for whom these assistants are most dangerous are the ones who cannot evaluate the answer, and the whole of your ability to use one safely rests on the understanding you built before you were allowed to lean on it.

## The limits that do not go away {#sec-remaining-limits}

Finishing the tool is a good moment to be clear-eyed about what your careful work does not buy you, and @sec-surviving-ai treats this at length. Three things are worth holding in mind as you close the project.

Your tests cover the cases you thought of, and there are always cases you did not think of, so the tests are a floor under your confidence and not a proof. Your own understanding of each piece is what stands behind them, which is why the rule about never accepting code you cannot read matters more at the end of a project than at the start. And no test you write about the code can catch a mistake in the biology the code assumes, because a flawless implementation of a wrong premise passes every test you can devise. Domain facts get checked against domain sources, not against the assistant that stated them.

Above all, responsibility stayed with you the whole way. You are submitting this tool, and if someone points at any function in it and asks how it works, the answer has to be yours.
